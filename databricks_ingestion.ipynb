```python
import json
import ssl
from kafka import KafkaConsumer
from datetime import datetime, timezone
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
from pyspark.sql.types import *


KAFKA_BOOTSTRAP_SERVERS = "wikimedia-kafka-khalilchikhi018-9584.c.aivencloud.com:13171"
KAFKA_TOPIC = "wikimedia-recentchange"
DELTA_TABLE_PATH = "/tmp/delta/wikimedia_events"
MAX_MESSAGES = 5000 
CONSUMER_TIMEOUT = 30000  

CA_FILE = "/tmp/ca.pem"
CERT_FILE = "/tmp/service.cert"
KEY_FILE = "/tmp/service.key"


SCHEMA = StructType([
    StructField("post_id", StringType(), True),
    StructField("type", StringType(), True),
    StructField("title", StringType(), True),
    StructField("namespace", IntegerType(), True),
    StructField("comment", StringType(), True),
    StructField("timestamp", LongType(), True),
    StructField("user", StringType(), True),
    StructField("bot", BooleanType(), True),
    StructField("server_name", StringType(), True),
    StructField("wiki", StringType(), True),
])


print(f"Connecting to Kafka at {KAFKA_BOOTSTRAP_SERVERS}...")

consumer = KafkaConsumer(
    KAFKA_TOPIC,
    bootstrap_servers=KAFKA_BOOTSTRAP_SERVERS,
    security_protocol="SSL",
    ssl_cafile=CA_FILE,
    ssl_certfile=CERT_FILE,
    ssl_keyfile=KEY_FILE,
    value_deserializer=lambda v: json.loads(v.decode("utf-8")),
    auto_offset_reset="earliest",
    consumer_timeout_ms=CONSUMER_TIMEOUT,
    group_id=None
)

records = []
for msg in consumer:
    records.append(msg.value)
    if len(records) >= MAX_MESSAGES:
        break

consumer.close()
print(f"Consumed {len(records)} records from Kafka.")

spark = SparkSession.builder.getOrCreate()

if not records:
    print("No new records. Exiting.")
else:
    flat_records = []
    for r in records:
        try:
            flat_records.append({
                "event_id": str(r.get("id", "")),
                "event_type": r.get("type", None),
                "page_title": r.get("title", None),
                "namespace": int(r.get("namespace", 0)),
                "editor": r.get("user", None),
                "is_bot": bool(r.get("bot", False)),
                "wiki": r.get("wiki", None),
                "server_name": r.get("server_name", None),
                "comment": str(r.get("comment", ""))[:500],
                "timestamp": int(r.get("timestamp", 0)),
            })
        except Exception:
            continue

    print(f"Flattened {len(flat_records)} records.")

    df = spark.createDataFrame(flat_records)

    enriched = (
        df
        .withColumn("event_time", F.to_timestamp(F.col("timestamp")))
        .withColumn("processed_at", F.current_timestamp())
        .withColumn(
            "editor_type",
            F.when(F.col("is_bot") == True, "bot").otherwise("human")
        )
        .filter(F.col("event_time").isNotNull())
        .filter(F.col("event_id") != "")
        .dropDuplicates(["event_id"])
    )

    print(f"Writing {enriched.count()} enriched records to Delta Lake...")

    (
        enriched.write
        .format("delta")
        .mode("append")
        .option("mergeSchema", "true")
        .saveAsTable("wikimedia_events")
    )

    print("Done. Sample:")
    spark.sql("SELECT * FROM wikimedia_events LIMIT 5").show(truncate=False)
```
